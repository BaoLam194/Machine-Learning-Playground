After we have the model build from previous training, we can load few images from test case to see how it is real correct!

```python
import numpy as np 
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers # type: ignore
from sklearn.model_selection import train_test_split

# Load data into variable
data = np.load("data/mnist.npz") 

# Extract train, test, file
x_train, x_test, y_train, y_test = data["x_train"], data["x_test"], data["y_train"], data["y_test"]
```

```python
# Load pretrained model
# Change the model here for testing
model = keras.models.load_model("final_model/CNNmodelver1.keras")

plt.figure(figsize=(14,8)) #width 10, height 10
indices = np.random.choice(len(x_test), 12, replace=False) # random choose 12 sample from test which no replace
print(x_test[indices].shape)
preds = model.predict(x_test[indices])   # shape (10, 10)
predicted_labels = np.argmax(preds, axis=1) # take max vector of each predict
for i, idx in enumerate(indices):
    plt.subplot(3, 4, i + 1)
    plt.imshow(x_test[idx], cmap="gray")  # change cmap if RGB
    plt.title(f"Correct: {y_test[idx]}, Predicted:{predicted_labels[i]}")

    plt.axis("off")
plt.tight_layout() 
plt.show()
```
